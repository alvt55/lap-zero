```python
import gymnasium as gym
from tqdm import tqdm 
import statistics
from stable_baselines3.common.monitor import Monitor
import os 
```

```python
log_dir = "./random_logs/"
os.makedirs(log_dir, exist_ok=True)

total_timesteps = 1_000_000
total_reward = 0 

episode_over = False
```

```python
# create env 
env = gym.make("CarRacing-v3", render_mode=None, lap_complete_percent=0.2, domain_randomize=False, continuous=False)

env = Monitor(env, log_dir)

# action space and observation space 
print(f"Action space: {env.action_space}")   # discrete 0-5 per action 
print(f"Observation space: {env.observation_space}") # value of 0 to 255, for 96x96 pixels 

# new episode 
observation, info = env.reset()
print(f"Sample action: {env.action_space.sample()}")  
print(f"Sample observation: {env.observation_space.sample()}") 
```

```python
for i in range(total_timesteps): 
    action = env.action_space.sample()  

    observation, reward, terminated, truncated, info = env.step(action)

    total_reward += reward

    episode_over = terminated or truncated

    if episode_over: 
        observation, info = env.reset()

print(f"Episode finished! Total reward: {total_reward}")
env.close()
```
